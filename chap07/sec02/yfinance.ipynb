```python
import yfinance as yf

# Microsoft(MSFT)에 대한 Ticker 객체 생성
msft=yf.Ticker("MSFT")

# Ticker 객체의 정보 출력(.py에서 실행할 때는 print(msft.info)로 사용)
display(msft.info)
```

```python
hist=msft.history(period="5d") # 5일간의 주가 데이터 가져오기
display(hist)
```
